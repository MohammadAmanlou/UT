# Vehicle Speed Check

## **Description**

The project uses Haar Cascade for detecting cars and Dlib’s correlation tracker for continuous tracking across frames. Every 10 frames, the system detects new cars, and their positions are tracked. The speed of each car is calculated based on the distance it travels between consecutive frames. This speed is displayed as text on the video feed. 
---

## **Requirements**
- **Libraries**: OpenCV, Dlib, NumPy
- **File Structure**:
  - Model files (e.g., `myhaar.xml`) are located in the `models/` directory.
  - Video files (e.g., `cars.mp4`) are located in the `videos/` directory.
  - Output video is saved in the current directory as `outpy.avi`.

---

## **Outputs**
1. **Video**:
   - Original video with detected cars tracked and their speeds displayed.

### **Sample Outputs**
- Processed video saved as `outpy.avi` with bounding boxes and speed annotations.

---

## **Algorithm Details**

1. **Car Detection**:
   - A Haar Cascade model is used to detect cars in the video frames. Detection occurs every 10 frames to identify new cars.

2. **Tracking with Dlib**:
   - Dlib’s `correlation_tracker` is used to continuously track the detected cars. A tracker is created for each detected car, and its position is updated in subsequent frames.

3. **Speed Estimation**:
   - The speed is estimated by calculating the distance between a car’s position in two consecutive frames and converting the distance from pixels to meters. The speed is displayed on the video feed in km/hr.

4. **Frame Processing**:
   - The system tracks multiple cars and removes trackers for cars no longer detected. It updates the positions of the remaining tracked cars and recalculates their speeds as they move across the frame.

5. **Output Generation**:
   - The processed video is saved and displayed, with bounding boxes around tracked cars and speed annotations displayed on the screen.

# Requirements

```python
import cv2
import dlib
import time
import math
import os
```

# Loading Model

```python
# Define the path to the Haar Cascade model file using os.path.join for cross-platform compatibility
MODEL_ADDRESS = os.path.join("models", "myhaar.xml")

# Define the path to the video file using os.path.join for cross-platform compatibility
VIDEO_ADDRESS = os.path.join("videos", "cars.mp4")

# Load the Haar Cascade model for car detection
carCascade = cv2.CascadeClassifier(MODEL_ADDRESS) # TODO

# Initialize video capture with the specified video file path
video = cv2.VideoCapture(VIDEO_ADDRESS) # TODO

# Constants for video frame dimensions
WIDTH = 1280  # Width of the video frame
HEIGHT = 720  # Height of the video frame
```

# Speed Detection

```python
# Function to estimate the speed of an object based on two locations
def estimateSpeed(location1, location2):
    d_pixels = math.sqrt(
        math.pow(location2[0] - location1[0], 2)
        + math.pow(location2[1] - location1[1], 2)
    )
    ppm = 8.8  # Pixels per meter
    d_meters = d_pixels / ppm  # Distance in meters
    fps = 18  # Frames per second
    speed = d_meters * fps * 3.6  # Speed in km/hr
    return speed

# Function to track multiple objects in a video
def trackMultipleObjects():
    rectangleColor = (0, 255, 0)  # Color of the rectangle around tracked objects
    frameCounter = 0  # Counter for frames
    currentCarID = 0  # ID for the current car being tracked
    fps = 0  # Frames per second

    carTracker = {}  # Dictionary to store car trackers
    carLocation1 = {}  # Dictionary to store initial car locations
    carLocation2 = {}  # Dictionary to store current car locations
    speed = [None] * 1000  # List to store speeds of cars

    out = cv2.VideoWriter(
        "outpy.avi", cv2.VideoWriter_fourcc("M", "J", "P", "G"), 10, (WIDTH, HEIGHT)
    )  # Video writer to save output video

    while True:
        start_time = time.time()  # Record the start time for FPS calculation
        rc, image = video.read()  # Read a frame from the video
        if type(image) == type(None):
            break

        image = cv2.resize(image, (WIDTH, HEIGHT))  # Resize the frame to desired dimensions
        resultImage = image.copy()  # Make a copy of the frame for result

        frameCounter = frameCounter + 1  # Increment frame counter

        carIDtoDelete = []  # List to store IDs of cars to delete

        # Update the trackers and check their quality
        for carID in carTracker.keys():
            trackingQuality = carTracker[carID].update(image)
            if trackingQuality < 7:
                carIDtoDelete.append(carID)

        # Remove cars that are no longer tracked
        for carID in carIDtoDelete:
            print("Removing carID " + str(carID) + " from list of trackers.")
            print("Removing carID " + str(carID) + " previous location.")
            print("Removing carID " + str(carID) + " current location.")
            carTracker.pop(carID, None)
            carLocation1.pop(carID, None)
            carLocation2.pop(carID, None)

        # Detect new cars every 10 frames
        if not (frameCounter % 10):
            gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
            cars = carCascade.detectMultiScale(gray, 1.1, 13, 18, (24, 24))
            for _x, _y, _w, _h in cars:
                x = int(_x)
                y = int(_y)
                w = int(_w)
                h = int(_h)
                x_bar = x + 0.5 * w
                y_bar = y + 0.5 * h
                matchCarID = None

                # Check if the detected car matches an existing car
                for carID in carTracker.keys():
                    trackedPosition = carTracker[carID].get_position()
                    t_x = int(trackedPosition.left())
                    t_y = int(trackedPosition.top())
                    t_w = int(trackedPosition.width())
                    t_h = int(trackedPosition.height())
                    t_x_bar = t_x + 0.5 * t_w
                    t_y_bar = t_y + 0.5 * t_h
                    if (
                        (t_x <= x_bar <= (t_x + t_w))
                        and (t_y <= y_bar <= (t_y + t_h))
                        and (x <= t_x_bar <= (x + w))
                        and (y <= t_y_bar <= (y + h))
                    ):
                        matchCarID = carID

                # If no match is found, create a new tracker
                if matchCarID is None:
                    print("Creating new tracker " + str(currentCarID))
                    tracker = dlib.correlation_tracker()
                    tracker.start_track(image, dlib.rectangle(x, y, x + w, y + h))
                    carTracker[currentCarID] = tracker
                    carLocation1[currentCarID] = [x, y, w, h]
                    currentCarID = currentCarID + 1

        # Update the positions of tracked cars
        for carID in carTracker.keys():
            trackedPosition = carTracker[carID].get_position()
            t_x = int(trackedPosition.left())
            t_y = int(trackedPosition.top())
            t_w = int(trackedPosition.width())
            t_h = int(trackedPosition.height())
            cv2.rectangle(
                resultImage, (t_x, t_y), (t_x + t_w, t_y + t_h), rectangleColor, 4
            )
            carLocation2[carID] = [t_x, t_y, t_w, t_h]

        end_time = time.time()  # Record the end time for FPS calculation
        if not (end_time == start_time):
            fps = 1.0 / (end_time - start_time)  # Calculate FPS

        # Calculate the speed of the cars
        for i in carLocation1.keys():
            if frameCounter % 1 == 0:
                [x1, y1, w1, h1] = carLocation1[i]
                [x2, y2, w2, h2] = carLocation2[i]
                carLocation1[i] = [x2, y2, w2, h2]
                if [x1, y1, w1, h1] != [x2, y2, w2, h2]:
                    if (speed[i] == None or speed[i] == 0) and y1 >= 275 and y1 <= 285:
                        speed[i] = estimateSpeed([x1, y1, w1, h1], [x2, y2, w2, h2])
                    if speed[i] != None and y1 >= 180:
                        cv2.putText(
                            resultImage,
                            str(int(speed[i])) + " km/hr",
                            (int(x1 + w1 / 2), int(y1 - 5)),
                            cv2.FONT_HERSHEY_SIMPLEX,
                            0.75,
                            (255, 255, 255),
                            2,
                        )

        # Display the result
        cv2.imshow("result", resultImage)
        if cv2.waitKey(33) == 27:
            break

    cv2.destroyAllWindows()

# Main function to start tracking
if __name__ == "__main__":
    trackMultipleObjects()
```

### Detailed Report on Car Detection, Tracking, and Speed Estimation System

#### Overview

This Python program implements a real-time system for detecting, tracking, and estimating the speed of cars in a video stream. The system utilizes OpenCV’s Haar Cascade classifier for initial car detection and dlib’s correlation tracker for tracking multiple vehicles across video frames. Speed estimation is calculated based on the change in position of tracked vehicles over time.

---

#### Key Components and Workflow

1. **Cross-Platform Path Definitions**

   * The program uses `os.path.join()` to define paths for the Haar Cascade model XML file and the input video file, ensuring compatibility across Windows, Linux, and macOS platforms.
   * `MODEL_ADDRESS` points to `"models/myhaar.xml"` (the pre-trained Haar Cascade for cars).
   * `VIDEO_ADDRESS` points to `"videos/cars.mp4"` (the input traffic video).

2. **Haar Cascade Loading and Video Capture**

   * The Haar Cascade classifier is loaded using OpenCV’s `cv2.CascadeClassifier()` for initial car detection.
   * Video input is opened via `cv2.VideoCapture()`, enabling frame-by-frame processing.
   * Frame dimensions are set as constants: width = 1280 pixels and height = 720 pixels, which standardize the input size for consistent processing.

3. **Speed Estimation Function**

   * `estimateSpeed(location1, location2)` calculates the speed of a vehicle given two positions in consecutive frames.
   * It computes the Euclidean distance in pixels between two locations.
   * Converts pixels to meters using a fixed scale (`ppm = 8.8` pixels per meter).
   * Uses the frame rate (`fps = 18`) to estimate speed in meters per second and converts it to kilometers per hour by multiplying by 3.6.
   * This method assumes a known fixed relationship between pixel distance and real-world distance, which depends on camera calibration.

4. **Multi-Object Tracking**

   * The main tracking happens inside the `trackMultipleObjects()` function.

   * **Initialization:**

     * Several dictionaries keep track of active car trackers, previous and current positions, and estimated speeds.
     * `carTracker` stores dlib correlation tracker objects for each car.
     * `carLocation1` and `carLocation2` hold bounding box positions at different frames.
     * A `frameCounter` helps manage when to perform detection vs. tracking.
     * A video writer `out` is initialized to save the processed output.

   * **Frame Processing Loop:**

     * Each frame is read, resized to fixed dimensions, and copied for drawing results.
     * The tracker dictionary is updated to maintain tracking quality:

       * If tracker quality falls below a threshold (7), the tracker and associated data are removed.
     * Every 10 frames, Haar Cascade detection runs on the grayscale frame to detect new cars.

       * For each detected car, the program checks if it matches an existing tracker by comparing bounding box overlaps.
       * If no match is found, a new tracker is created with dlib's `correlation_tracker` and added to `carTracker`.
     * All tracked cars have rectangles drawn around them on the result image.
     * FPS (frames per second) is dynamically calculated based on processing time for performance monitoring.

   * **Speed Calculation:**

     * For each tracked car, the code compares the current and previous locations.
     * When the vehicle crosses a certain Y-axis band (between 275 and 285 pixels), speed estimation is triggered.
     * The estimated speed is displayed above the tracked vehicle’s bounding box in km/hr.

   * **Output Display:**

     * The processed frames are shown in a window named `"result"`.
     * The loop continues until the user presses the 'Esc' key (`cv2.waitKey(33) == 27`).
     * After the loop, OpenCV windows are closed properly.

---

#### Technical Strengths and Design Choices

* **Combination of Detection and Tracking:**
  Using Haar Cascade every 10 frames reduces computational load, while continuous tracking by dlib maintains smooth tracking between detections.

* **Robust Tracker Management:**
  The system removes trackers with poor quality, preventing drift and inaccurate tracking over time.

* **Speed Estimation Based on Real-World Scaling:**
  By defining pixels-per-meter (ppm) and frame rate, the system converts pixel displacement into meaningful speed measurements.

* **Modular Structure:**
  Functions are clearly separated (`estimateSpeed()` and `trackMultipleObjects()`), which improves readability and maintainability.

* **Video Output (Optional):**
  Although the code initializes a video writer to save the output, frames are not explicitly written in the given snippet but can be added easily.

---

#### Potential Improvements and Considerations

* **Calibration Needed for Accuracy:**
  The pixels-per-meter constant is hardcoded and may require calibration for different camera setups or perspectives.

* **Variable Frame Rate:**
  Assuming a fixed frame rate might cause inaccuracies if the actual video FPS differs.

* **Performance Optimization:**
  For higher-resolution videos or more vehicles, the system might require optimization or hardware acceleration.

* **False Positives / Negatives:**
  Haar Cascade can sometimes produce false detections or miss vehicles, suggesting that integrating deep learning-based detectors could improve accuracy.

* **Enhanced Tracking:**
  Incorporating additional data association techniques (e.g., Kalman filters, SORT algorithm) could improve multi-object tracking reliability.
