# Library-Based Image Processing Implementation

## 📥 Project Overview

I started with an input image in JPG format. Using ImageIO, I loaded the image directly into Python. The goal was to:

- **Decompose** the image into three different color models by extracting individual color components.
- **Generate** additional image versions: a grayscale image and a binary image.
- **Compute** and plot histograms for both the grayscale image and the original RGB channels.
- **Smooth** the grayscale histogram using a moving average technique.
- **Perform** a Fourier Transform on the grayscale image.
- **Save** all output images as JPG files and display them for visual analysis.

Throughout the project, I leveraged powerful libraries to handle image I/O, color conversions, histogram computations, and Fourier analysis with minimal manual overhead. 😄💻

---

## 🔍 Step-by-Step Methodology

### 1. Image Loading and Preprocessing
- **Input Acquisition:**  
  I used `imageio.imread("input_image.jpg")` to load the input image in its original JPG format.  
- **Grayscale Conversion:**  
  I manually converted the image to grayscale using the standard luminosity method by applying weights (0.2989, 0.5870, 0.1140) to the R, G, and B channels respectively. This is crucial for further processing like histogram analysis and Fourier transformation.  
- **Binary Image Creation:**  
  The grayscale image was thresholded at 127 to produce a binary (black and white) image, enabling me to observe high-contrast features.

### 2. Color Component Decomposition into 3 Models
I decomposed the image into the following three color representations:

- **RGB Model:**  
  I extracted the Red, Green, and Blue channels individually. Each channel was saved as a grayscale image, which clearly shows the intensity distribution of that particular color component.  
  - *Example:* The red channel image highlights how much red is present in different regions of the original image.

- **HSV Model:**  
  I converted the image to the HSV color space using OpenCV's `cv2.cvtColor` function. This provided three new components:  
  - **H (Hue):** Indicates the dominant color.
  - **S (Saturation):** Represents the vibrancy of the color.
  - **V (Value):** Reflects the brightness level.
  Saving these channels helps in understanding the color composition in terms of human-perceived attributes.

- **CMY Model:**  
  Since CMY is the complementary color model, I computed these channels by subtracting the RGB components from 255. This model provides insight into the inverse color composition of the image.
  
All the extracted channels (R, G, B, H, S, V, C, M, Y) were saved as separate output images for further examination.

### 3. Histogram Analysis and Contrast Evaluation
- **Grayscale Histogram:**  
  I computed the histogram of the grayscale image using NumPy’s histogram functions to count the frequency of pixel intensities (0–255). The plot shows how the pixel intensity values are distributed, which is a direct indicator of the image’s contrast.  
  - **Analysis:**  
    A well-contrasted image will have a histogram that is spread over a wide range of intensities, whereas a low-contrast image will have a histogram clustered around a narrow range of intensities.

- **RGB Histogram:**  
  Histograms were also computed for each of the RGB channels. These overlaid histograms (red, green, and blue lines) help in understanding the color dynamics and balance.  
  - **Analysis:**  
    Comparing the RGB histogram with the grayscale histogram can reveal differences in contrast between the color and luminance representations.

- **Smoothed Grayscale Histogram:**  
  To reduce noise and better visualize the underlying distribution, I smoothed the grayscale histogram using a moving average filter (kernel size of 5). This provides a clearer view of the major intensity trends in the image.  
  - **Analysis:**  
    The smoothed histogram allows for improved contrast analysis by reducing minor fluctuations.

### 4. Fourier Transform Analysis
- **Fourier Transformation:**  
  I applied a 2D Fourier Transform to the grayscale image using NumPy's FFT functions. The result was shifted to center the zero frequency components and the magnitude was computed.  
- **Visualization:**  
  The magnitude spectrum was scaled and saved as an image.  
  - **Analysis:**  
    The Fourier spectrum helps in identifying periodic patterns and frequency components present in the image. High-frequency components indicate abrupt transitions (such as edges), and low-frequency components relate to the overall structure. This transform is particularly useful for analyzing texture and smooth transitions in the image.

### 5. Final Output and Visualization
- **Saving Outputs:**  
  Every intermediate result (color-decomposed images, grayscale, binary, histograms, Fourier transform) was saved as a JPG file in the designated output folder.
- **Display:**  
  Finally, using Matplotlib, I arranged all images and plots into a multi-panel figure to display them simultaneously. This comprehensive visual output allows for direct comparison between the different processing steps and models, facilitating a detailed analysis.

```python
import imageio
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

os.makedirs("output_library", exist_ok=True)
img_color = imageio.imread("input_image.jpg")
img_gray = (np.dot(img_color[...,:3], [0.2989, 0.5870, 0.1140])).astype(np.uint8)
img_binary = ((img_gray > 127) * 255).astype(np.uint8)
R = img_color[:,:,0]
G = img_color[:,:,1]
B = img_color[:,:,2]
R_img = np.stack([R, R, R], axis=2)
G_img = np.stack([G, G, G], axis=2)
B_img = np.stack([B, B, B], axis=2)
hsv = cv2.cvtColor(img_color, cv2.COLOR_RGB2HSV)
H = hsv[:,:,0]
S = hsv[:,:,1]
V = hsv[:,:,2]
H_img = np.stack([H, H, H], axis=2)
S_img = np.stack([S, S, S], axis=2)
V_img = np.stack([V, V, V], axis=2)
cmy = 255 - img_color
C = cmy[:,:,0]
M = cmy[:,:,1]
Y = cmy[:,:,2]
C_img = np.stack([C, C, C], axis=2)
M_img = np.stack([M, M, M], axis=2)
Y_img = np.stack([Y, Y, Y], axis=2)
imageio.imwrite("output_library/color.jpg", img_color)
imageio.imwrite("output_library/grayscale.jpg", img_gray)
imageio.imwrite("output_library/binary.jpg", img_binary)
imageio.imwrite("output_library/R.jpg", R_img)
imageio.imwrite("output_library/G.jpg", G_img)
imageio.imwrite("output_library/B.jpg", B_img)
imageio.imwrite("output_library/H.jpg", H_img)
imageio.imwrite("output_library/S.jpg", S_img)
imageio.imwrite("output_library/V.jpg", V_img)
imageio.imwrite("output_library/C.jpg", C_img)
imageio.imwrite("output_library/M.jpg", M_img)
imageio.imwrite("output_library/Y.jpg", Y_img)
bins = np.arange(256)
hist_gray, _ = np.histogram(img_gray.flatten(), bins=256, range=(0,256))
fig1, ax1 = plt.subplots()
ax1.plot(bins, hist_gray)
ax1.set_title("Grayscale Histogram")
fig1.savefig("output_library/histogram_grayscale.jpg")
hist_R, _ = np.histogram(R.flatten(), bins=256, range=(0,256))
hist_G, _ = np.histogram(G.flatten(), bins=256, range=(0,256))
hist_B, _ = np.histogram(B.flatten(), bins=256, range=(0,256))
fig2, ax2 = plt.subplots()
ax2.plot(bins, hist_R, color='r')
ax2.plot(bins, hist_G, color='g')
ax2.plot(bins, hist_B, color='b')
ax2.set_title("RGB Histogram")
fig2.savefig("output_library/histogram_rgb.jpg")
kernel = np.ones(5) / 5.0
hist_gray_smoothed = np.convolve(hist_gray, kernel, mode='same')
fig3, ax3 = plt.subplots()
ax3.plot(bins, hist_gray_smoothed)
ax3.set_title("Smoothed Grayscale Histogram")
fig3.savefig("output_library/histogram_gray_smoothed.jpg")
F = np.fft.fft2(img_gray)
F_shift = np.fft.fftshift(F)
magnitude = 20 * np.log(np.abs(F_shift) + 1)
magnitude_norm = 255 * (magnitude - magnitude.min()) / (magnitude.max() - magnitude.min())
fourier_img = magnitude_norm.astype(np.uint8)
imageio.imwrite("output_library/fourier.jpg", fourier_img)
fig4, ax4 = plt.subplots()
ax4.imshow(fourier_img, cmap="gray")
ax4.set_title("Fourier Transform")
plt.tight_layout()

figA, axsA = plt.subplots(4, 3, figsize=(15,15))
axsA[0,0].imshow(img_color)
axsA[0,0].set_title("Color")
axsA[0,1].imshow(img_gray, cmap="gray")
axsA[0,1].set_title("Grayscale")
axsA[0,2].imshow(img_binary, cmap="gray")
axsA[0,2].set_title("Binary")
axsA[1,0].imshow(R_img, cmap="gray")
axsA[1,0].set_title("R Channel")
axsA[1,1].imshow(G_img, cmap="gray")
axsA[1,1].set_title("G Channel")
axsA[1,2].imshow(B_img, cmap="gray")
axsA[1,2].set_title("B Channel")
axsA[2,0].imshow(H_img, cmap="gray")
axsA[2,0].set_title("H Channel")
axsA[2,1].imshow(S_img, cmap="gray")
axsA[2,1].set_title("S Channel")
axsA[2,2].imshow(V_img, cmap="gray")
axsA[2,2].set_title("V Channel")
axsA[3,0].imshow(C_img, cmap="gray")
axsA[3,0].set_title("C Channel")
axsA[3,1].imshow(M_img, cmap="gray")
axsA[3,1].set_title("M Channel")
axsA[3,2].imshow(Y_img, cmap="gray")
axsA[3,2].set_title("Y Channel")
for ax in axsA.flat:
    ax.axis('off')
plt.show()
plt.show()
```

## 📊 Analysis of the Results

- **Color Conversion:**  
  The original image, when converted to grayscale and binary forms, reveals that the weighted sum method successfully captures the luminance information. The binary image clearly demarcates regions of high contrast, indicating that the chosen threshold is effective.

- **Color Decomposition:**  
  The RGB channels clearly isolate the intensity distribution of each primary color. The manual HSV conversion produces an H channel that effectively delineates the dominant hues, while the S and V channels offer insight into the saturation and brightness, respectively. The CMY channels further complement this by showing the inverse color relationships.

- **Histogram Analysis:**  
  The manually computed grayscale histogram indicates a wide spread of intensity values, which correlates with the overall contrast in the image. The overlaid RGB histograms reveal how each color contributes to the overall tone. The smoothed grayscale histogram, displayed using a vibrant color map, successfully reduces noise and highlights the main intensity peaks, aiding in contrast evaluation.

- **Fourier Transform:**  
  The Fourier transform visualization clearly shows the frequency components. The central region, corresponding to low frequencies, is prominent, while the high-frequency components at the edges indicate the presence of fine details and sharp transitions within the image. This transform serves as a critical tool for texture and edge analysis.



```python
import os
import numpy as np
import imageio
import matplotlib.pyplot as plt

os.makedirs("output", exist_ok=True)
img = imageio.imread("input_image.jpg")

gray = (0.2989 * img[..., 0] + 0.5870 * img[..., 1] + 0.1140 * img[..., 2]).astype(np.uint8)

binary = (gray > 127) * 255
binary = binary.astype(np.uint8)

R = img[:,:,0]
G = img[:,:,1]
B = img[:,:,2]
R_img = np.stack([R, R, R], axis=2)
G_img = np.stack([G, G, G], axis=2)
B_img = np.stack([B, B, B], axis=2)

def rgb_to_hsv_manual(rgb):
    rgb = rgb.astype(np.float32) / 255.0
    R = rgb[:,:,0]
    G = rgb[:,:,1]
    B = rgb[:,:,2]
    maxc = np.maximum(np.maximum(R, G), B)
    minc = np.minimum(np.minimum(R, G), B)
    V = maxc
    delta = maxc - minc
    S = np.zeros_like(maxc)
    mask = maxc > 0
    S[mask] = delta[mask] / maxc[mask]
    H = np.zeros_like(maxc)
    mask_r = (maxc == R) & (delta != 0)
    H[mask_r] = ((G[mask_r] - B[mask_r]) / delta[mask_r])
    mask_g = (maxc == G) & (delta != 0)
    H[mask_g] = ((B[mask_g] - R[mask_g]) / delta[mask_g]) + 2
    mask_b = (maxc == B) & (delta != 0)
    H[mask_b] = ((R[mask_b] - G[mask_b]) / delta[mask_b]) + 4
    H = (H * 60) % 360
    H = (H / 360 * 255).astype(np.uint8)
    S = (S * 255).astype(np.uint8)
    V = (V * 255).astype(np.uint8)
    hsv = np.stack([H, S, V], axis=2)
    return hsv

hsv = rgb_to_hsv_manual(img)
H_channel = hsv[:,:,0]
S_channel = hsv[:,:,1]
V_channel = hsv[:,:,2]
H_img = np.stack([H_channel, H_channel, H_channel], axis=2)
S_img = np.stack([S_channel, S_channel, S_channel], axis=2)
V_img = np.stack([V_channel, V_channel, V_channel], axis=2)

cmy = 255 - img
C = cmy[:,:,0]
M = cmy[:,:,1]
Y = cmy[:,:,2]
C_img = np.stack([C, C, C], axis=2)
M_img = np.stack([M, M, M], axis=2)
Y_img = np.stack([Y, Y, Y], axis=2)

imageio.imwrite("output/color.jpg", img)
imageio.imwrite("output/grayscale.jpg", gray)
imageio.imwrite("output/binary.jpg", binary)
imageio.imwrite("output/R.jpg", R_img)
imageio.imwrite("output/G.jpg", G_img)
imageio.imwrite("output/B.jpg", B_img)
imageio.imwrite("output/H.jpg", H_img)
imageio.imwrite("output/S.jpg", S_img)
imageio.imwrite("output/V.jpg", V_img)
imageio.imwrite("output/C.jpg", C_img)
imageio.imwrite("output/M.jpg", M_img)
imageio.imwrite("output/Y.jpg", Y_img)

hist_gray = np.zeros(256, dtype=int)
for pixel in gray.flatten():
    hist_gray[pixel] += 1

hist_R = np.zeros(256, dtype=int)
hist_G = np.zeros(256, dtype=int)
hist_B = np.zeros(256, dtype=int)
for pixel in R.flatten():
    hist_R[pixel] += 1
for pixel in G.flatten():
    hist_G[pixel] += 1
for pixel in B.flatten():
    hist_B[pixel] += 1

kernel_size = 5
hist_gray_smooth = np.zeros(256, dtype=float)
half_kernel = kernel_size // 2
for i in range(256):
    s = 0.0
    count = 0
    for j in range(i - half_kernel, i + half_kernel + 1):
        if j >= 0 and j < 256:
            s += hist_gray[j]
            count += 1
    hist_gray_smooth[i] = s / count

F = np.fft.fft2(gray)
F_shift = np.fft.fftshift(F)
magnitude = 20 * np.log(np.abs(F_shift) + 1)
magnitude_norm = 255 * (magnitude - magnitude.min()) / (magnitude.max() - magnitude.min())
fourier_img = magnitude_norm.astype(np.uint8)
imageio.imwrite("output/fourier.jpg", fourier_img)
fig_all, axs = plt.subplots(6, 3, figsize=(15, 30))

axs[0,0].imshow(img)
axs[0,0].set_title("Color Image")
axs[0,1].imshow(gray, cmap="gray")
axs[0,1].set_title("Grayscale Image")
axs[0,2].imshow(binary, cmap="gray")
axs[0,2].set_title("Binary Image")

axs[1,0].imshow(R_img, cmap="gray")
axs[1,0].set_title("R Channel")
axs[1,1].imshow(G_img, cmap="gray")
axs[1,1].set_title("G Channel")
axs[1,2].imshow(B_img, cmap="gray")
axs[1,2].set_title("B Channel")
axs[2,0].imshow(H_img, cmap="gray")
axs[2,0].set_title("H Channel (HSV)")
axs[2,1].imshow(S_img, cmap="gray")
axs[2,1].set_title("S Channel (HSV)")
axs[2,2].imshow(V_img, cmap="gray")
axs[2,2].set_title("V Channel (HSV)")

axs[3,0].imshow(C_img, cmap="gray")
axs[3,0].set_title("C Channel (CMY)")
axs[3,1].imshow(M_img, cmap="gray")
axs[3,1].set_title("M Channel (CMY)")
axs[3,2].imshow(Y_img, cmap="gray")
axs[3,2].set_title("Y Channel (CMY)")

axs[4,0].bar(np.arange(256), hist_gray, color="gray")
axs[4,0].set_title("Grayscale Histogram (Manual)")
axs[4,1].plot(np.arange(256), hist_R, color="r", label="R")
axs[4,1].plot(np.arange(256), hist_G, color="g", label="G")
axs[4,1].plot(np.arange(256), hist_B, color="b", label="B")
axs[4,1].legend()
axs[4,1].set_title("RGB Histogram (Manual)")
axs[4,2].bar(np.arange(256), hist_gray_smooth)
axs[4,2].set_title("Smoothed Grayscale Histogram (Manual)")

axs[5,0].imshow(fourier_img,cmap = "gray")
axs[5,0].set_title("Fourier Transform (Grayscale)")
axs[5,1].axis("off")
axs[5,2].axis("off")

for ax in axs.flat:
    ax.axis("off")
plt.tight_layout()
plt.show()
```
