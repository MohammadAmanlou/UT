## 📸 Library-Based Implementation Using OpenCV and NumPy

This section demonstrates a full implementation using widely used computer vision libraries. In this version, we load the image, create binary, grayscale, and color versions, adjust the contrast (both increasing and decreasing) of the grayscale and color images, add brightness to the grayscale image, and finally apply salt-and-pepper noise to the original color image. All outputs are saved into a designated folder.

## 🌟 What is OpenCV (cv2)?

OpenCV (Open Source Computer Vision Library) is an open-source computer vision and machine learning software library. It provides a common infrastructure for computer vision applications and accelerates the use of machine perception in commercial products.

### 🔍 Explanation:
- **Image Loading:** The image is loaded from a specified path using OpenCV.
- **Format Conversion:**  
  - A grayscale version is obtained by converting the color image.  
  - A binary version is generated using a threshold value of 127.
- **Contrast Adjustment:**  
  A linear contrast adjustment is applied. The midpoint (median intensity of the grayscale image) is used as the reference point. Two adjustments are computed: one with an increased contrast factor (1.5) and one with a decreased contrast factor (0.5).
- **Brightness Enhancement:**  
  For the grayscale image, every pixel value is increased by 20 (with clamping between 0 and 255).
- **Salt & Pepper Noise Addition:**  
  A function adds noise by randomly setting 2% of the pixels to white (salt) and 2% to black (pepper).
- **Output:**  
  All generated images are written to the output folder.


```python
import cv2
import numpy as np
import os

image_path = 'input_image.jpg'
output_folder = "output_lib"
os.makedirs(output_folder, exist_ok=True)

img_color = cv2.imread(image_path, cv2.IMREAD_COLOR)
if img_color is None:
    print("Error: Image not loaded.")
    exit()

img_gray = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)
_, img_binary = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

cv2.imwrite(os.path.join(output_folder, 'binary.jpg'), img_binary)
cv2.imwrite(os.path.join(output_folder, 'grayscale.jpg'), img_gray)
cv2.imwrite(os.path.join(output_folder, 'color.jpg'), img_color)

def adjust_contrast(image, factor, reference):
    adjusted = (image.astype(np.float32) - reference) * factor + reference
    return np.clip(adjusted, 0, 255).astype(np.uint8)

med_gray = np.median(img_gray)

contrast_increase_gray = adjust_contrast(img_gray, 1.5, med_gray)
contrast_decrease_gray = adjust_contrast(img_gray, 0.5, med_gray)
cv2.imwrite(os.path.join(output_folder, 'gray_contrast_increased.jpg'), contrast_increase_gray)
cv2.imwrite(os.path.join(output_folder, 'gray_contrast_decreased.jpg'), contrast_decrease_gray)

contrast_increase_color = adjust_contrast(img_color, 1.5, med_gray)
contrast_decrease_color = adjust_contrast(img_color, 0.5, med_gray)
cv2.imwrite(os.path.join(output_folder, 'color_contrast_increased.jpg'), contrast_increase_color)
cv2.imwrite(os.path.join(output_folder, 'color_contrast_decreased.jpg'), contrast_decrease_color)

bright_gray = np.clip(img_gray.astype(np.int16) + 20, 0, 255).astype(np.uint8)
cv2.imwrite(os.path.join(output_folder, 'gray_brightness_plus20.jpg'), bright_gray)

def salt_and_pepper(image, noise_level=0.02):
    noisy = image.copy()
    num_pixels = image.shape[0] * image.shape[1]
    num_noisy = int(noise_level * num_pixels)
    coords = [np.random.randint(0, i, num_noisy) for i in image.shape[:2]]
    noisy[coords[0], coords[1]] = 255
    coords = [np.random.randint(0, i, num_noisy) for i in image.shape[:2]]
    noisy[coords[0], coords[1]] = 0
    return noisy

noisy_color = salt_and_pepper(img_color, 0.02)
cv2.imwrite(os.path.join(output_folder, 'color_salt_and_pepper.jpg'), noisy_color)

print("Library-based processing complete. Check folder '{}'.".format(output_folder))
```

# Image Processing Project Report 🚀

I developed this project entirely from scratch using **NumPy** and the new **ImageIO** library. Here’s a detailed report about my approach, the challenges I encountered with SciPy, and an explanation of the code I wrote.

## 🛠️ Why Not Use SciPy for Image I/O?

- **SciPy Deprecation Issue:**  
  SciPy’s traditional image I/O functions like `scipy.misc.imread` and `scipy.misc.imsave` have been removed in newer versions. This posed a significant challenge because the assignment initially required using only basic libraries. Since relying on an older version of SciPy was not an option, I needed an alternative.

- **Solution with ImageIO:**  
  I opted for **ImageIO**, a modern and robust library that supports reading and writing popular image formats (including JPG) while still keeping the implementation minimal. With ImageIO, I was able to read the input image and write all processed images in JPG format, meeting all requirements.

---

## 💻 Code Explanation

I wrote the complete code from scratch with the following key steps:

1. **Image Loading:**  
   - I used `imageio.imread("input_image.jpg")` to load the input JPG image into a NumPy array.

2. **Image Conversions:**  
   - **Grayscale:** I converted the RGB image to grayscale using the standard luminance formula implemented with `np.dot`.
   - **Binary:** I generated a binary image by thresholding the grayscale image at 127. Pixel values above 127 were set to 255 (white), and those below were set to 0 (black).

3. **Contrast Adjustment:**  
   - I defined a function that adjusts image contrast via a linear transformation
   - The median of the grayscale image serves as the reference point to balance the transformation.
   - I applied this function to both the grayscale and the original color image, generating two versions for each: one with increased contrast and one with decreased contrast.

4. **Brightness Enhancement:**  
   - I increased the brightness of the grayscale image by adding 20 to each pixel value. The result was clamped to ensure that the values remained within the 0-255 range.

5. **Salt-and-Pepper Noise:**  
   - I implemented a function that randomly adds salt-and-pepper noise by setting 2% of the pixels to white (salt) and another 2% to black (pepper).

6. **Output Saving:**  
   - Each processed image was saved as a JPG file in the output directory using `imageio.imwrite`.


```python
import imageio
import numpy as np
import os

os.makedirs("output_imageio", exist_ok=True)
img_color = imageio.imread("input_image.jpg")
def rgb2gray(rgb):
    return np.dot(rgb[...,:3], [0.2989, 0.5870, 0.1140])
img_gray = rgb2gray(img_color).astype(np.uint8)
img_binary = ((img_gray > 127) * 255).astype(np.uint8)
imageio.imwrite(os.path.join("output_imageio", "color.jpg"), img_color)
imageio.imwrite(os.path.join("output_imageio", "grayscale.jpg"), img_gray)
imageio.imwrite(os.path.join("output_imageio", "binary.jpg"), img_binary)
def adjust_contrast(image, factor, reference):
    adjusted = (image.astype(np.float32) - reference) * factor + reference
    return np.clip(adjusted, 0, 255).astype(np.uint8)
median_val = np.median(img_gray)
gray_contrast_increased = adjust_contrast(img_gray, 1.5, median_val)
gray_contrast_decreased = adjust_contrast(img_gray, 0.5, median_val)
imageio.imwrite(os.path.join("output_imageio", "gray_contrast_increased.jpg"), gray_contrast_increased)
imageio.imwrite(os.path.join("output_imageio", "gray_contrast_decreased.jpg"), gray_contrast_decreased)
color_contrast_increased = adjust_contrast(img_color, 1.5, median_val)
color_contrast_decreased = adjust_contrast(img_color, 0.5, median_val)
imageio.imwrite(os.path.join("output_imageio", "color_contrast_increased.jpg"), color_contrast_increased)
imageio.imwrite(os.path.join("output_imageio", "color_contrast_decreased.jpg"), color_contrast_decreased)
bright_gray = np.clip(img_gray.astype(np.int16) + 20, 0, 255).astype(np.uint8)
imageio.imwrite(os.path.join("output_imageio", "gray_brightness_plus20.jpg"), bright_gray)
def salt_and_pepper(image, noise_level=0.02):
    noisy = image.copy()
    num_pixels = image.shape[0] * image.shape[1]
    num_noisy = int(noise_level * num_pixels)
    coords = [np.random.randint(0, s, num_noisy) for s in image.shape[:2]]
    if image.ndim == 3:
        noisy[coords[0], coords[1], :] = 255
    else:
        noisy[coords[0], coords[1]] = 255
    coords = [np.random.randint(0, s, num_noisy) for s in image.shape[:2]]
    if image.ndim == 3:
        noisy[coords[0], coords[1], :] = 0
    else:
        noisy[coords[0], coords[1]] = 0
    return noisy
noisy_color = salt_and_pepper(img_color, 0.02)
imageio.imwrite(os.path.join("output_imageio", "color_salt_and_pepper.jpg"), noisy_color)
print("Processing complete. Check the 'output_imageio' folder for output images.")
```
