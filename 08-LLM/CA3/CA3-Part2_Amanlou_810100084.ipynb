## 1. An Overview of Information Retrieval (IR) and RAG (2 points)


- **Information Retrieval (IR)**: The process of obtaining information system resources relevant to a specific information need from a collection of those resources. Each IR system consists of a collection of documents, a set of queries, and a retrieval function that ranks the documents based on their relevance to the query.
- **Retrieval-Augmented Generation (RAG)**: A model that combines the strengths of retrieval-based and generation-based approaches. It retrieves relevant documents from a large corpus and uses them to generate a response to a query. RAG is particularly useful for tasks where the answer is not explicitly present in the training data but can be inferred from related documents.
- **RAG Architecture**: The RAG architecture consists of two main components:
  - **Retriever**: This component retrieves relevant documents from a large corpus based on the input query. It can be implemented using various retrieval methods, such as BM25 or dense retrieval.
  - **Generator**: This component generates a response based on the retrieved documents and the input query. It can be implemented using transformer-based models.

**Temperature** controls the randomness of the output. Lower values make the text more focused and deterministic, while higher values allow for more creativity but can sometimes lead to less coherent responses.

**Max length** (or `max_new_tokens`) sets a limit on the number of tokens the model can generate, ensuring that the output doesn’t become overly long.

**Top-p** (nucleus sampling) determines the cumulative probability of token selection. The model only considers tokens within a certain probability range, like 90% for `top_p=0.9`, balancing randomness and coherence.

**Top-k** restricts the model to choosing from the top-k most likely tokens at each step. For example, `top_k=50` lets the model pick from the 50 most probable options. Smaller values reduce randomness, while larger values increase it.

**Repetition penalty** discourages the model from repeating words or phrases excessively by penalizing tokens it has already generated. This helps avoid repetitive or looping text.

```python
from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser

human_message = HumanMessagePromptTemplate.from_template(
    "Tell me some brief information about the football player named {person_name}."
)

prompt = ChatPromptTemplate.from_messages([human_message])

output_parser = StrOutputParser()

class SimpleChain:
    def __init__(self, llm, prompt, output_parser):
        self.llm = llm
        self.prompt = prompt
        self.output_parser = output_parser

    def invoke(self, input_dict):
        messages = self.prompt.format_messages(**input_dict)
        input_text = messages[0].content

        # استفاده از متد invoke بجای __call__
        output = self.llm.invoke(input_text)

        # اگر خروجی رشته است مستقیم پارس کن
        # اگر خروجی دیکشنری یا لیست بود، باید اصلاح شود
        if isinstance(output, str):
            raw_text = output
        elif isinstance(output, list) and isinstance(output[0], dict) and 'generated_text' in output[0]:
            raw_text = output[0]['generated_text']
        else:
            # اگر فرمت خروجی متفاوت است اینجا اصلاح کن
            raw_text = str(output)

        return self.output_parser.parse(raw_text)


simple_chain = SimpleChain(llm=llm, prompt=prompt, output_parser=output_parser)

```

```python
from langchain_core.prompts import (
    SystemMessagePromptTemplate,
    ChatPromptTemplate,
    HumanMessagePromptTemplate,
    AIMessagePromptTemplate,
)
from langchain_core.output_parsers import JsonOutputParser

system_message = SystemMessagePromptTemplate.from_template(
    "You are a helpful assistant that provides detailed info about football players in JSON format."
)

human_message = HumanMessagePromptTemplate.from_template(
    (
        "Provide the following information about the football player named {player_name}:\n"
        "- full name\n"
        "- nationality\n"
        "- age\n"
        "- current club\n"
        "- position\n\n"
        "Format your response strictly as a JSON object."
    )
)

# Optional: AI message template just for formatting (can be omitted)
ai_message = AIMessagePromptTemplate.from_template("{answer}")

# Build chat prompt with system, human (and AI) messages
prompt = ChatPromptTemplate.from_messages([system_message, human_message])

# Use JsonOutputParser to parse JSON response into dict
output_parser = JsonOutputParser()

class JsonChain:
    def __init__(self, llm, prompt, output_parser):
        self.llm = llm
        self.prompt = prompt
        self.output_parser = output_parser

    def invoke(self, input_dict):
        messages = self.prompt.format_messages(**input_dict)
        input_text = messages[1].content  # human message content (index 1 is human prompt)
        output = self.llm.invoke(input_text)
        return self.output_parser.parse(output)

    def batch(self, input_list):
        results = []
        for input_dict in input_list:
            try:
                result = self.invoke(input_dict)
            except Exception as e:
                result = {"error": str(e), "input": input_dict}
            results.append(result)
        return results

# Instantiate the chain
json_chain = JsonChain(llm=llm, prompt=prompt, output_parser=output_parser)
```

### 3.2 Sparse Retriever (3 points)

In this section, we would create a sparse retriever for our RAG pipeline.

#### Question 5: (2 points)

1. Explain how a sparse retriever like TF-IDF represents documents and queries. How does this representation influence which documents are retrieved?
2. Sparse retrievers rely on exact token matches between queries and documents. What are the strengths and weaknesses of this approach, especially compared to dense retrievers?

TF-IDF is a sparse retrieval method that represents documents and queries using high-dimensional sparse vectors, where each dimension corresponds to a unique term in the vocabulary. These vectors are created using the TF-IDF formula, which assigns a weight to each term based on its importance. The term frequency (TF) captures how often a term appears in a specific document or query, while the inverse document frequency (IDF) measures how rare or common the term is across all documents, giving higher weight to rarer terms. Since most documents and queries contain only a small subset of the entire vocabulary, the resulting vectors are mostly zeros, making them sparse.
This sparsity makes retrieval more efficient, as calculating similarity between the query vector and document vectors—typically using cosine similarity—requires fewer computations due to the large number of zero entries. Documents that share many important terms (i.e., terms with high TF-IDF scores) with the query are assigned higher similarity scores, while those without overlapping or significant terms rank lower.
However, TF-IDF's reliance on exact term matching means it works best when the query and relevant documents use the same or very similar vocabulary. It struggles with vocabulary mismatch, such as synonyms, potentially missing relevant documents when different wording is used. Despite this limitation, TF-IDF remains efficient, interpretable, and effective for many retrieval tasks where exact word overlap plays a key role.

**Strengths of Sparse Retrievers (like TF-IDF):**

* **Interpretability:** Easy to understand why documents are retrieved—based on exact term matches and their importance.
* **Efficiency:** Fast indexing and retrieval using inverted indices, suitable for very large corpora.
* **Simplicity:** Requires no training; straightforward to implement and tune.
* **Good for keyword-based queries:** Works well when the query and documents share exact words.

---

**Weaknesses of Sparse Retrievers:**

* **Vocabulary mismatch:** Cannot handle synonyms, paraphrases, or related concepts well since it relies on exact token matching.
* **Limited semantic understanding:** Fails when relevant documents use different wording from the query.
* **Sparse representation limitations:** Cannot capture the contextual meaning or deeper relationships between words.

---

**Compared to Dense Retrievers:**

* **Dense retrievers** use continuous vector embeddings capturing semantic similarity, allowing them to match related concepts even if the exact words differ.
* Dense methods better handle paraphrasing, synonyms, and more natural language queries.
* However, dense retrievers typically require expensive training, more computation for indexing and retrieval, and are less interpretable.

### 3.3 Semantic Retriever (4 points)

#### Question 6: (2 point)

1. How does a semantic retriever represent documents and queries differently from a sparse retriever? Explain why this helps in cases where there are no exact word overlaps.
2. What role do embedding models (e.g., sentence-transformers) play in semantic retrieval? How does the choice of embedding model affect the retriever’s performance?

1. Semantic retrievers represent documents and queries using dense vector embeddings that capture the meaning and context of the text, rather than just counting exact word occurrences. Each document and query is converted into a continuous vector in a high-dimensional space, where similar meanings are placed close together, even if they don't share the same exact words.

This helps because when queries and documents use different wording but have similar meanings (e.g., synonyms, paraphrases), semantic retrievers can still match them based on their closeness in the embedding space. In contrast, sparse retrievers rely on exact token matches and fail when there's no direct word overlap.

2. Embedding models transform sentences or documents into dense vectors that encode their semantic content. Models like sentence-transformers are specially trained to produce embeddings that place semantically similar texts close together in the vector space.

The choice of embedding model greatly affects retriever performance because:

Quality of semantic understanding: Better models capture deeper context, nuances, and relationships, improving retrieval accuracy.

Domain adaptation: Models trained or fine-tuned on specific domains perform better on related queries/documents.

Vector dimensionality and efficiency: A balance between embedding size and retrieval speed impacts system scalability.

In short, a good embedding model enables semantic retrievers to find relevant documents even without exact word matches by understanding the meaning behind the text.

2. Describe two different strategies for integrating retrieved context into the prompt. What are the trade-offs between these approaches?

1. **Main Components of a RAG System and Their Interaction During Inference:**

* **Retriever:**
  Receives the user input (query) and searches a large document collection to find the most relevant documents or passages related to the query. This can be a sparse retriever (like TF-IDF) or a dense retriever (using embeddings).

* **Generator:**
  A language model that takes the retrieved documents along with the original query and generates a natural language response. The generator uses the retrieved context to produce more informed, accurate, or up-to-date answers.

**Flow during inference:**
User inputs a query → Retriever fetches relevant documents → Retrieved documents and query are combined into a prompt → Generator produces the final output based on this enriched prompt → Output is returned to the user.

---

2. **Two Strategies for Integrating Retrieved Context into the Prompt and Their Trade-offs:**

* **Concatenation (Prepending the Context):**
  The retrieved documents are directly prepended or appended to the user query to form a single prompt fed into the language model.

  * *Pros:* Simple to implement; leverages the language model’s ability to read context.
  * *Cons:* Limited by prompt length; too much context can overwhelm the model or truncate important information.

* **Fusion-in-Decoder (FiD):**
  Retrieved documents are encoded separately and the decoder generates the answer by attending to all encoded documents jointly.

  * *Pros:* Allows handling many documents without prompt length limits; better at combining multiple pieces of information.
  * *Cons:* More complex architecture and higher computational cost.

#### Cache-Augmented Generation (CAG): (4 points)

1. What is Cache-Augmented Generation (CAG)? How does it improve efficiency or performance during generation?

Cache-Augmented Generation, or CAG, represents a significant optimization technique for modern language models that makes text generation more efficient by intelligently reusing computational work rather than performing redundant calculations. At its heart, CAG operates on a simple but powerful principle: why recompute what you've already figured out? When generating text word by word or token by token, language models often perform many similar calculations repeatedly, especially in their attention mechanisms where the same contextual information gets processed multiple times.  

The process works by storing intermediate results—like the key-value pairs in transformer models' attention layers—in a cache during text generation. As the model produces each new token, it can refer back to this cache rather than recalculating everything from scratch. This approach dramatically cuts down on unnecessary computation, making the entire generation process faster and more resource-efficient. For example, when generating a long response, the early parts of the text that have already been processed don’t need their attention states recomputed for every subsequent token; instead, the cached representations can be reused, saving both time and computational power.  

One of the most noticeable benefits of CAG is speed. Since the model isn't bogged down by redundant calculations, it can generate text more quickly, which is especially valuable in real-time applications like chatbots or interactive AI systems where low latency is crucial. Additionally, because caching reduces the computational load, it allows models to handle longer sequences without running into memory bottlenecks. This is particularly important for applications that require maintaining large contexts, such as document summarization or extended conversations, where processing lengthy inputs efficiently is essential.  

Beyond speed, CAG also improves resource efficiency. By minimizing redundant operations, it reduces the strain on GPU memory and processing power, making it more practical to deploy large language models in cost-sensitive or resource-constrained environments. This efficiency can translate into lower operational costs, better scalability, and the ability to run high-performance models on less powerful hardware.  

In practice, CAG is one of several optimizations that help make modern AI systems viable for real-world use. Whether in AI assistants, automated content generation, or other applications, the ability to generate text quickly and efficiently while maintaining high quality is a game-changer. By smartly reusing past computations rather than starting from zero at each step, Cache-Augmented Generation ensures that language models can deliver responsive, coherent, and context-aware outputs without unnecessary computational overhead. This makes CAG a key enabler for scalable, high-performance AI systems that need to balance speed, cost, and accuracy.

Cache-Augmented Generation (CAG) and Retrieval-Augmented Generation (RAG) are both techniques designed to enhance how language models generate text, but they approach the problem from fundamentally different angles. Understanding their similarities and differences helps determine when each method is most useful.

Both CAG and RAG share the broader goal of improving text generation beyond what a base language model can do alone. They operate on the principle that augmentation—whether through internal optimizations or external knowledge—can lead to better outcomes. CAG achieves this by reusing previously computed model states during generation, while RAG pulls in relevant documents or data from external sources to inform the model’s responses. In this sense, both methods rely on some form of additional information beyond the immediate input to enhance performance. However, the nature of that information and how it’s applied differ significantly.

The key distinction lies in their primary objectives. CAG is all about efficiency and speed. By caching intermediate computations—such as attention keys and values in transformer models—it avoids redundant calculations during text generation. This makes the process faster, reduces memory usage, and allows the model to handle longer sequences without excessive overhead. For example, in a chatbot that generates multi-turn conversations, CAG ensures that previously processed context isn’t recomputed from scratch at each step, leading to quicker responses. It doesn’t change the model’s knowledge or reasoning—just how efficiently it accesses what it already knows.

RAG, on the other hand, prioritizes knowledge and accuracy. Instead of optimizing internal computations, it dynamically retrieves relevant documents or data from an external knowledge base—such as Wikipedia, a proprietary database, or real-time web sources—and uses them to ground the model’s responses. This means the model can provide answers that are more factual, up-to-date, or domain-specific than what its base training data alone would allow. For instance, a customer support bot using RAG could pull the latest product manuals to answer user queries accurately, even if those details weren’t part of its original training.

Choosing between CAG and RAG depends on the specific needs of the application. If the priority is low latency, cost efficiency, or handling long contexts smoothly, CAG is the better choice. It’s ideal for real-time systems where speed matters, or where the base model’s knowledge is sufficient but needs to be accessed more efficiently. Meanwhile, RAG shines when factual correctness, access to dynamic knowledge, or domain expertise is critical. It’s the go-to method for question-answering systems, research assistants, or any scenario where responses must be grounded in external, verifiable information.

Interestingly, these techniques aren’t mutually exclusive. In some advanced systems, they can be combined—using CAG to accelerate the generation process while RAG ensures the content is accurate and informed by external data. This hybrid approach exemplifies how modern AI systems leverage multiple optimizations to balance speed, cost, and quality.

In summary, CAG and RAG represent two complementary paths to enhancing language models: one by making generation faster and leaner, the other by making it smarter and more informed. The right choice depends on whether the bottleneck is computational efficiency or knowledge depth—and in many cutting-edge applications, the answer might be both.

Models like CLIP (Contrastive Language–Image Pretraining) are designed to bridge the gap between text and images by embedding both into a shared vector space. This means that whether you input a photo of a dog or the sentence "a golden retriever playing fetch," CLIP can map them to nearby points in a high-dimensional space where their meanings align.  

The magic happens through a dual-encoder architecture—one neural network processes images (often a Vision Transformer or CNN), while another handles text (typically a Transformer). These networks are trained together on massive datasets of paired images and captions, using a contrastive learning objective. Essentially, the model learns to pull matching image-text pairs closer in the vector space while pushing unrelated pairs apart. For example, a picture of a sunset and its caption "vibrant sunset over the ocean" would end up with similar embeddings, while a mismatched pair like the same sunset image paired with "a busy city street" would be far apart.  

A key advantage of this shared space is its versatility in cross-modal retrieval, especially for Retrieval-Augmented Generation (RAG) systems. Imagine a search engine where you could type "waterfall in a tropical forest" and get relevant images, or upload a photo of a dish and retrieve recipes that match it—all because both modalities "speak the same language" in the embedding space. This eliminates the need for separate systems to handle text and images, streamlining infrastructure and enabling more intuitive multimodal interactions.  

However, the approach isn’t without challenges. Training CLIP effectively requires enormous datasets of high-quality image-text pairs, and the alignment between modalities isn’t always perfect. A painting of a cat might embed near the text "feline portrait," but subtle nuances—like the artist’s style or emotional tone—could be lost. There’s also a computational cost: running two encoders and comparing high-dimensional vectors demands significant resources. And while the shared space sacrifices some modality-specific specialization, the tradeoff is often worth it for applications needing flexible cross-modal understanding.  

In practice, CLIP-style embeddings have unlocked powerful use cases, from improving accessibility (e.g., generating alt text for images) to enhancing creative tools (like text-to-image search). But they’re not a silver bullet—designers must weigh the benefits of unified embeddings against their limitations, ensuring the system aligns with real-world needs. When done well, though, this technique brings us closer to AI that truly understands the connections between words and visuals.

When dealing with document images in systems like Colpali, breaking them down into smaller patches before processing isn't just a technical detail—it fundamentally changes how well the system understands and retrieves information. Imagine trying to analyze a dense research paper or a business report all at once; you'd likely miss important details buried in the layout. That's exactly why patch-based processing makes such a difference in multimodal RAG systems.

By splitting documents into manageable patches—whether they're paragraphs, tables, diagrams, or captions—the system can examine each section with precision. This is especially valuable because documents aren't uniform; a single page might contain a sidebar with key statistics, an illustrative chart, and blocks of explanatory text. Patch-level analysis preserves this structural nuance, allowing the model to recognize that a query about "sales growth trends" should prioritize the line graph patch rather than the methodology paragraph nearby.

This granular approach solves several real-world problems. First, it prevents small but critical elements from getting drowned out when processing full pages—a footnote with crucial disclaimers or a finely printed data table won't be overlooked. Second, it maintains the spatial relationships between components, so the system understands that a caption belongs to its adjacent image or that a bulleted list organizes related concepts. When you later search for content, this means results can pinpoint exact relevant sections rather than vaguely matching entire documents.

The benefits extend to performance too. Since patches isolate content regions, noisy elements like watermarks or marginalia won't distort the representation of important content. The system also gains efficiency; instead of laboriously reprocessing whole documents for every query, it can selectively compare query-relevant patches, speeding up retrieval. In applications like legal document review or academic research, where precision matters, this patch-aware retrieval can mean the difference between surfacing the perfect reference or drowning in vaguely related results.

Of course, there are challenges—determining optimal patch sizes, handling overlapping content, and ensuring seamless reconstruction of the full document context. But when implemented well, as in Colpali's approach, patch-based processing transforms document images from blurry snapshots into searchable, semantically rich maps where every element has its place and purpose. This is why modern multimodal RAG systems increasingly adopt this strategy for handling real-world documents where layout carries as much meaning as the raw text itself.

When building retrieval-augmented generation (RAG) systems that need to handle both text and images—or other non-text data like audio or video—one approach is to convert everything into text before retrieval. This might seem counterintuitive at first: why turn a rich image into a description when you could process it directly? But there are some compelling reasons this strategy works well in practice—along with some tradeoffs worth understanding.  

On the upside, using text as a common language simplifies everything. Retrieval systems have spent decades optimizing how to search through text efficiently, with tools like keyword matching, semantic search, and dense vector retrieval. By converting an image into a caption (e.g., "a red bicycle leaning against a brick wall"), you can leverage these battle-tested methods without reinventing the wheel. This unification also makes the system more interpretable: if a user searches for "bicycle photos" and gets a result, it’s easy to trace back that the match happened because the system stored the textual description "red bicycle." Storage becomes cheaper too—a few kilobytes of text replace megabytes of image data—and indexing scales more smoothly.  

Text also acts as a universal bridge. Imagine a system that needs to retrieve medical scans, satellite images, and podcast clips alongside research papers. Describing all of them in text (e.g., "MRI showing a torn meniscus," "aerial view of suburban sprawl," "interview about renewable energy policy") lets a single retrieval pipeline handle wildly different formats without needing custom architectures for each.  

But the drawbacks are real. Text is a lossy format for visuals. A painting described as "abstract blue and yellow shapes" loses the emotional impact of brushstrokes; a flowchart converted to text might miss the clarity of its arrows and spatial layout. These losses compound if the conversion step is weak—say, an image-captioning model misidentifies a dog as a wolf, or OCR garbles a critical number in a table. Even with perfect conversions, text struggles to capture subjective qualities like ambiance, artistic style, or the humor in a meme. Retrieval might surface technically correct but irrelevant results (e.g., matching "cat on a couch" to a cartoon when the user wanted real photos). There’s also added latency and cost from running separate models to generate descriptions upfront.  

So when does this approach make sense? If your use case prioritizes broad accessibility, simplicity, and integration with existing NLP tools—and if losing some visual nuance is acceptable—textual conversion is a pragmatic choice. But for applications where pixel-perfect accuracy or rich multimodal interaction matters (e.g., design platforms, medical imaging), hybrid systems that preserve native modalities often work better. The key is aligning the strategy with what users actually need to retrieve and why.
