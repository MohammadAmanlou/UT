### ✂️ 2.4 Temporal Train-Validation-Test Split and Normalization  
- **Temporal Split:** For each individual time series (e.g., each asset or symbol), the data is split **sequentially** into three parts:  
  - The **earliest portion** is used for **training**  
  - The **middle portion** for **validation**  
  - The **latest portion** for **testing**  
  This approach respects the chronological order of the data and follows the ratio defined in the reference paper. Unlike random splits, **temporal splits** ensure that future data is never used to inform the past.

- **Normalization:**  
  The features are normalized using a **Min-Max Scaler** that is **fitted only on the training set**. The same scaler is then applied to the validation and test sets to prevent data leakage.
